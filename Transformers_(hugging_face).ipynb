```python
#Text summarization
reader = pipeline ("summarization")
Output = reader ( text,max_length = 50,clean_up_tokenization_spaces = True)
print (Output [0]['summary_text'])
```

```python
#Translation
translator = pipeline("translation_en_to_de",model ="Helsinki-NLP/opus-mt-en-sw")
output = translator(text,clean_up_tokenization_spaces = True,max_length = 100)
print(output[0]['translation_text'])
```

```python
#Text generation
generator = pipeline("text-generation")
response = "Dear customer,we were sorry for the delay of your refund,our customer services will reach you soon"
prompt = text + "\n\n customer care:\n"+ response
output = generator(prompt,max_length = 100)
print(output[0]['generated_text'])
```
