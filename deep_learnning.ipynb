```python
from tensorflow import keras
from tensorflow.keras import layers

from tensorflow.keras.datasets import mnist

(train_data,train_label),_= mnist.load_data()

train_data = train_data.reshape(60000,28*28)

train_data = train_data.astype("float32")/255

my_model = keras.Sequential()

my_model.add(layers.Dense(512,activation = "relu"))
my_model.add(layers.Dense(10,activation = "softmax"))
my_model.compile(optimizer = "RMSprop",
                 metrics = ["accuracy"],
                 loss = "sparse_categorical_crossentropy")

my_model.fit(train_data,train_label, epochs= 10,batch_size =128, validation_split = 0.2)
```

```python
import matplotlib.pyplot as plt
val_loss = my_model.history.history["val_loss"]

epochs = range(1, len(val_loss) + 1)

plt.plot(epochs,val_loss,"b--",label = "validation loss")

plt.title('validation loss')
plt.xlabel("epochs")
plt.ylabel("validation_loss")
plt.legend()
```

```python
from tensorflow import keras
import numpy as fg
from tensorflow.keras.datasets import imdb
(train_data,train_label),_ = imdb.load_data(num_words = 10000)

def vectorize_sequences(sequences, dimensions = 10000):
    results = fg.zeros((len(sequences), dimensions))
    for i , sequence in enumerate(sequences):
        results [i, sequence] = 1

    return results

train_data = vectorize_sequences(train_data)
my_model = keras.Sequential()

my_model.add(layers.Dense(16,activation = "relu"))
my_model.add(layers.Dense(16,activation = "relu"))
my_model.add(layers.Dense(1,activation = "sigmoid"))
my_model.compile(optimizer = "RMSprop",
                 metrics = ["accuracy"],
                 loss = "binary_crossentropy")

history_original = my_model.fit(train_data,train_label, epochs= 20,batch_size =512, validation_split = 0.4)
```

```python
from tensorflow import keras
from tensorflow.keras import regularizers
my_model = keras.Sequential()

my_model.add(layers.Dense(16, kernel_regularizer = regularizers.l2(0.02),activation = "relu"))
my_model.add(layers.Dense(16,kernel_regularizer = regularizers.l2(0.02),activation = "relu"))
my_model.add(layers.Dense(1,activation = "sigmoid"))
my_model.compile(optimizer = "RMSprop",
                 metrics = ["accuracy"],
                 loss = "binary_crossentropy")

history_regularized = my_model.fit(train_data,train_label, epochs= 20,batch_size =512, validation_split = 0.4)
```

```python
import matplotlib.pyplot as plt
val_loss = history_original.history["val_loss"]
val_loss_reg = history_regularized.history["val_loss"]

epochs = range(1, len(val_loss) + 1)

plt.plot(epochs,val_loss,"b--",label = "validation loss")
plt.plot(epochs,val_loss_reg,"b-",label = "validation loss of regularized model")

plt.title('validation loss')
plt.xlabel("epochs")
plt.ylabel("validation_loss")
plt.legend()
```

```python
from tensorflow import keras
import numpy as fg
from tensorflow.keras.datasets import imdb
(train_data,train_label),_ = imdb.load_data(num_words = 10000)

def vectorize_sequences(sequences, dimensions = 10000):
    results = fg.zeros((len(sequences), dimensions))
    for i , sequence in enumerate(sequences):
        results [i, sequence] = 1

    return results

train_data = vectorize_sequences(train_data)
my_model = keras.Sequential()

my_model.add(layers.Dense(16,activation = "relu"))
my_model.add(layers.Dense(16,activation = "relu"))

my_model.add(layers.Dense(1,activation = "sigmoid"))
my_model.compile(optimizer = "RMSprop",
                 metrics = ["accuracy"],
                 loss = "binary_crossentropy")

history_original = my_model.fit(train_data,train_label, epochs= 20,batch_size =512, validation_split = 0.4)
```

```python
from tensorflow import keras
from tensorflow.keras import regularizers
my_model = keras.Sequential()

my_model.add(layers.Dense(16,activation = "relu"))
my_model.add(layers.Dropout(0.5))
my_model.add(layers.Dense(16,activation = "relu"))
my_model.add(layers.Dropout(0.5))
my_model.add(layers.Dense(1,activation = "sigmoid"))
my_model.compile(optimizer = "RMSprop",
                 metrics = ["accuracy"],
                 loss = "binary_crossentropy")

history_regularized_dropout = my_model.fit(train_data,train_label, epochs= 20,batch_size =512, validation_split = 0.4)
```

```python
import matplotlib.pyplot as plt
val_loss = history_original.history["val_loss"]
val_loss_reg = history_regularized_dropout.history["val_loss"]

epochs = range(1, len(val_loss) + 1)

plt.plot(epochs,val_loss,"b--",label = "validation loss")
plt.plot(epochs,val_loss_reg,"b-",label = "validation loss of regularized model with dropout")

plt.title('validation loss')
plt.xlabel("epochs")
plt.ylabel("validation_loss")
plt.legend()
```
